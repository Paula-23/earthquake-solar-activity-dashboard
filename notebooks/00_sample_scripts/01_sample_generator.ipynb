```python
# ==============================================================================
# 2. ESTABLISH WORKING PATHS (CONVERSION FROM STEP 2)
# ==============================================================================
notebook_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(notebook_dir, "..", ".."))

DATA_DIR = os.path.join(project_root, "data")
RAW_DIR = os.path.join(DATA_DIR, "raw")

# Define exactly which source files map to which sample outputs
FILE_MAPPING = {
    "gfz": {
        "source": os.path.join(RAW_DIR, "gfz_api.json"),
        "target": os.path.join(DATA_DIR, "gfz_sample_raw.json")
    },
    "usgs": {
        "source": os.path.join(RAW_DIR, "usgs_eq_raw_S1.json"), # Updated to S1
        "target": os.path.join(DATA_DIR, "usgs_sample_raw.json")
    }
}

# Number of earthquake records you want to extract for your sample
SAMPLE_SIZE = 50
```

```python
# ==============================================================================
# 3. READ, SAMPLE, AND WRITE LOOP
# ==============================================================================
print("🚀 Starting the sampling process...\n")

for provider, paths in FILE_MAPPING.items():
    source_path = paths["source"]
    target_path = paths["target"]
    
    # A. Open and read the large source file
    with open(source_path, "r", encoding="utf-8") as file:
        original_data = json.load(file)

    # B. Locate the earthquake array inside the GeoJSON 'features' list
    all_events = original_data.get("features", [])
    total_events = len(all_events)

    # C. Safety Check: If the file happens to be smaller than our target sample size
    items_to_pick = min(SAMPLE_SIZE, total_events)

    # D. Grab a clean random selection of unique events
    random_sample_events = random.sample(all_events, items_to_pick)

    # E. Duplicate the metadata container wrapper, but replace the giant array
    sample_output = original_data.copy()
    sample_output["features"] = random_sample_events

    # F. Save the new small file into the main 'data/' directory
    with open(target_path, "w", encoding="utf-8") as file:
        json.dump(sample_output, file, indent=4, ensure_ascii=False)

    print(f"✅ Generated sample for {provider.upper()}:")
    print(f"   • Selected {items_to_pick} random events out of {total_events} available.")
    print(f"   • Saved to: {target_path}\n")

print("🎉 All done! Look inside your project's main 'data' folder to find the sample files.")
```
